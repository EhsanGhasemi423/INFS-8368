**Derivatives and Differentiation**

**What is a derivative?**

A derivative measures how quickly a function changes when its input changes.

Think of it as the slope of a curve at a particular point.

**Derivative = Instantaneous rate of change**


$$
f'(x)=\lim_{h\rightarrow0}\frac{f(x+h)-f(x)}{h}
$$

Change in output/change in input

It is called **the average rate of change**.



**Derivatives** can tell us how rapidly a loss function would increase or decrease were we to increase or decrease each parameter by an infinitesimally small amount.



A **differentiable** function has a well-defined slope at every point where it is differentiable.

**Differentiability** means a function has a well-defined derivative.

**Gradient-based learning** requires derivatives, so deep learning models optimize differentiable loss functions (surrogate losses)

Not every function has a derivative (e.g., Accuracy, AUC).

**"How fast is the function changing at x=1?"**

The blue curve doesn't directly answer this.
We need a way to measure steepness.

A curve doesn't have one constant slope.

So we temporarily replace the tiny neighborhood around x=1 with the best straight line.

That straight line is the tangent line.

Why?

Because a straight line has a single slope.

The slope of the tangent line is the derivative. at x = 1, the slope is 2. At x = 2 the slopw is 8.






```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return 3*x**2 - 4*x

x = np.linspace(0, 3, 100)

plt.figure(figsize=(6,4))
plt.plot(x, f(x), label='f(x)')
plt.plot(x, 2*x-3, '--', label='Tangent line at x=1')
plt.scatter(1, f(1), color='red')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()
```

## Derivative as a Tangent Line

The blue curve represents the function

$$
f(x)=3x^2-4x.
$$

The red point indicates the point where

$$
x=1.
$$

The derivative of the function is

$$
f'(x)=6x-4.
$$

Evaluating the derivative at \(x=1\),

$$
f'(1)=6(1)-4=2.
$$

The orange dashed line is the tangent line at \(x=1\).

### Key Idea

The derivative represents the slope of the tangent line at a specific point.

In this example,

$$
f'(1)=2.
$$

This means that **at \(x=1\), the function is increasing at a rate of 2.**

### Example: Find the Derivative

Given

$$
f(x)=3x^2-4x
$$

Using the derivative rules,

$$
\begin{aligned}
f'(x)
&=\frac{d}{dx}(3x^2-4x)\\
&=3\frac{d}{dx}(x^2)-4\frac{d}{dx}(x)\\
&=3(2x)-4(1)\\
&=6x-4.
\end{aligned}
$$

Therefore,

$$
\boxed{f'(x)=6x-4.}
$$
