## Deep Recurrent Neural Networks

So far, we have mainly considered an RNN with **one recurrent hidden layer**.

For more **complex sequence problems**, we may want the network to learn
**multiple levels of representation**.


A **Deep RNN** is created by stacking multiple recurrent layers.


### How Information Flows

At each time step $t$, information moves in **two directions**:

**1. Across time within the same layer**

$$
H_{t-1}^{(l)} \rightarrow H_t^{(l)}
$$

The layer uses information remembered from the previous time step.

**2. Up through the network**

$$
H_t^{(l-1)} \rightarrow H_t^{(l)}
$$

The output of one recurrent layer becomes the input to the next recurrent layer.




### Hidden State of a Deep RNN

For layer $l$ at time $t$:

$$
H_t^{(l)}
=
\phi_l
\left(
H_t^{(l-1)}W_{xh}^{(l)}
+
H_{t-1}^{(l)}W_{hh}^{(l)}
+
b_h^{(l)}
\right)
$$

where:

- $H_t^{(l)}$ = hidden state of **layer $l$ at the current time step**
- $H_t^{(l-1)}$ = output coming from the **layer below**
- $H_{t-1}^{(l)}$ = hidden state coming from the **previous time step**
- $W_{xh}^{(l)}$ = weights for information coming from the layer below
- $W_{hh}^{(l)}$ = weights for information coming from the previous time step
- $b_h^{(l)}$ = bias
- $\phi_l$ = activation function

For the **first recurrent layer**, the input from the layer below is simply the original input:

$$
H_t^{(0)} = X_t
$$

So conceptually:

$$
X_t
\rightarrow
H_t^{(1)}
\rightarrow
H_t^{(2)}
\rightarrow
\cdots
\rightarrow
H_t^{(L)}
$$

The output is then calculated from the **last recurrent layer**:

$$
O_t = H_t^{(L)}W_{hq} + b_q
$$



drnn.png

To calculate the current hidden state, a layer looks DOWN to the layer below and BACK to its own previous hidden state.

### Main Idea

A Deep RNN does not introduce a new type of memory mechanism.

It simply **stacks recurrent layers**:

**Input → RNN Layer 1 → RNN Layer 2 → ... → RNN Layer L → Output**

The recurrent layers can also be replaced with **LSTM or GRU layers**.

> **Horizontal direction = sequence/time**  
> **Vertical direction = network depth**
