## Encoder–Decoder Architecture

Machine translation is a **sequence-to-sequence (seq2seq)** problem:

> Given a sequence in a source language, generate a sequence in a target language.

The source and target sequences may have **different lengths** and are not necessarily aligned word by word.

A common solution is the **encoder–decoder architecture**:

**Source Sequence → Encoder → Encoded State → Decoder → Target Sequence**

### Encoder

The **encoder** reads the source sequence and transforms it into an internal representation called the **encoded state**.

Example:

`I am hungry . <eos>`

↓  

**Encoder**

↓  

**Encoded State**

The input sequence can have a **variable number of tokens**, while the encoded state has a **fixed shape** determined by the architecture.

For example, if an RNN encoder has 32 hidden units, its final hidden state contains 32 values regardless of whether the source contains 5, 10, or 20 tokens.

### Decoder

The **decoder** receives the encoded state and generates the target sequence **one token at a time**.

For example:

`<bos> → J' → ai → faim → . → <eos>`

At each step, the decoder uses:

- the encoded information from the source
- its current hidden state
- the previous target token

to predict the **next target token**.

The decoder continues generating tokens until it produces `<eos>`.

Therefore, the target sequence can have a **different length from the source sequence**.

## Putting the Encoder and Decoder Together

The complete sequence-to-sequence process is:

**Variable-length Source Sequence**

↓

**Encoder**

↓

**Fixed-shape Encoded State**

↓

**Decoder**

↓

**Variable-length Target Sequence**

For machine translation:

`I am hungry . <eos>`

↓ **Encoder**

`Encoded State`

↓ **Decoder**

`<bos> J' ai faim . <eos>`

The encoder and decoder therefore perform two different tasks:

1. **Encoder:** represents the source sequence.
2. **Decoder:** uses that representation to generate the target sequence.

Importantly, **fixed shape refers to the dimensionality of the encoded state, not the number of output tokens**.

The input and output sequence lengths can both vary:

**Variable-length input → Fixed-shape state → Variable-length output**

This encoder–decoder architecture forms the foundation for the **sequence-to-sequence RNN model** that we study next.

### From Architecture to a Seq2Seq Model

So far, we have described the **encoder–decoder architecture conceptually**:

**Source Sequence → Encoder → Encoded State → Decoder → Target Sequence**

But we have not yet specified **how the encoder and decoder actually perform these computations**.

Since we already know how RNNs process sequences, a natural approach is to use:

- an **RNN encoder** to process the source sequence and produce the encoded state;
- an **RNN decoder** to generate the target sequence one token at a time.

This gives us a **sequence-to-sequence (seq2seq) RNN**.
