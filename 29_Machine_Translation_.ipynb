# From Sequence Modeling to Machine Translation

So far, we have used recurrent neural networks to model **one sequence over time**.

For example, in language modeling, the model receives previous tokens and predicts what comes next:

$$
x_1 \rightarrow x_2 \rightarrow x_3 \rightarrow \cdots
$$

We also saw that bidirectional RNNs can use context from **both directions** when the entire input sequence is available.

Machine translation introduces a different problem.

Instead of predicting another token from the **same sequence**, we want to transform one complete sequence into a **different sequence**.




## Machine Translation as a Sequence-to-Sequence Problem

Example:

**Source sentence:**

`I am hungry.`

**Target sentence:**

`J'ai faim.`

This is called a **sequence-to-sequence (seq2seq)** problem:

$$
\text{Source sequence}
\rightarrow
\text{Target sequence}
$$

This is more challenging than ordinary language modeling because:

- the input and output are **different sequences**
- they may have **different lengths**
- they may use **different vocabularies**
- corresponding words may appear in a **different order**

So a single recurrent chain is no longer enough.

We need one part of the model to **understand the source sequence** and another part to **generate the target sequence**.

This motivates the:

$$
\boxed{\text{Encoder} \rightarrow \text{Decoder}}
$$

architecture.




## Preparing Translation Sequences

Before feeding text into the model, sentences are converted into **tokens**.

Example:

**Source**

`hi . <eos>`

**Target**

`<bos> salut . <eos>`

### Special Tokens

- `<bos>` — beginning of sequence
- `<eos>` — end of sequence
- `<pad>` — fills shorter sequences so examples in the same batch have equal length
- `<unk>` — represents an unknown or infrequent token

These tokens help the model understand where a sequence **starts**, where it **ends**, and how shorter sequences are handled in a minibatch.

For example:

`hi . <eos> <pad> <pad>`

`how are you ? <eos>`

The `<pad>` tokens do not represent meaningful words. They are added so that sequences can be processed efficiently in batches.





## Why This Matters

Machine translation gives us a natural reason to introduce a new architecture:

> **How can one network read an entire source sequence and another network generate a new target sequence from it?**

The answer is the **encoder–decoder architecture**, which we study next.
