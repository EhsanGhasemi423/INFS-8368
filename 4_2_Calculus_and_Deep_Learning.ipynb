## Partial Derivatives and Gradients

So far, we have worked with functions of **one variable**.

For example,

$$
f(x)=3x^2-4x
$$

The derivative

$$
f'(x)=\frac{df}{dx}
$$

tells us **how the function changes when we change one variable, $x$.**

---

### Deep Learning is Different

In deep learning, the loss does **not** depend on just one variable.

Instead, it depends on **many model parameters (weights)**.

For example,

$$
L(w_1,w_2,\ldots,w_n)
$$

where

- $L$ = Loss function
- $w_1,w_2,\ldots,w_n$ = model weights (parameters)

Now we ask a different question:

> **What happens to the loss if we change only one weight while keeping all the others fixed?**

---

### Partial Derivative

The answer is given by a **partial derivative**.

For example,

$$
\frac{\partial L}{\partial w_1}
$$

means:

> *How does the loss change if we change only weight $w_1$?*

Similarly,

$$
\frac{\partial L}{\partial w_2}
$$

asks:

> *How does the loss change if we change only weight $w_2$?*

Notice the symbol changes from

$$
\frac{d}{dx}
$$

to

$$
\frac{\partial}{\partial w_i}
$$

because the function now has **many variables**.

---

### Gradient

A neural network may have thousands or even millions of weights.

Instead of computing one derivative, we compute one partial derivative for **every weight**.

Collecting all of these partial derivatives forms the **gradient**:

$$
\nabla L=
\begin{bmatrix}
\frac{\partial L}{\partial w_1}\\
\frac{\partial L}{\partial w_2}\\
\vdots\\
\frac{\partial L}{\partial w_n}
\end{bmatrix}
$$

The gradient is simply a vector that contains **one partial derivative for each model parameter.**

---

### Why is the Gradient Important?

During training, TensorFlow automatically computes the gradient.

Gradient descent then uses it to update every weight:

$$
w_{\text{new}}
=
w_{\text{old}}
-
\eta
\frac{\partial L}{\partial w}
$$

where

- $\eta$ = learning rate
- $\frac{\partial L}{\partial w}$ = slope of the loss with respect to the weight

The goal is to adjust the weights so that the **loss becomes smaller** after each update.

---

### Key Takeaways

- **Derivative:** One variable → one slope.
- **Partial derivative:** Many variables → slope with respect to one variable.
- **Gradient:** A vector containing all partial derivatives.
- **Gradient descent:** Uses the gradient to update all model weights and minimize the loss.

**Chain Rule**

A neural network is a composition of many functions, and the chain rule lets TensorFlow compute gradients through all of them automatically.

## Chain Rule

The derivative rules we learned earlier are sufficient for simple functions such as

$$
f(x)=3x^2-4x.
$$

However, a neural network is **not** a single function.

Instead, it is a sequence of functions applied one after another.

For example,

$$
y=f(g(x)).
$$

Here,

- $g(x)$ transforms the input,
- $f(\cdot)$ transforms the output of $g(x)$.

To find how $y$ changes with respect to $x$, we must consider **both functions**.

The **chain rule** tells us how to do this:

$$
\frac{dy}{dx}
=
\frac{dy}{du}
\cdot
\frac{du}{dx},
$$

where

$$
u=g(x).
$$

In words,

> **The rate of change of the final output equals the product of the rates of change at each step.**

---

### Why is the Chain Rule Important?

A deep neural network is simply many functions connected together.

For example,

$$
x
\rightarrow
\text{Layer 1}
\rightarrow
\text{Layer 2}
\rightarrow
\text{Layer 3}
\rightarrow
L.
$$

To compute the gradient of the loss with respect to the weights, we repeatedly apply the chain rule from the output layer back to the input layer.

Fortunately, TensorFlow performs these calculations automatically using **automatic differentiation** (also called **backpropagation**).

Therefore, when training a neural network, we usually do **not** compute the chain rule by hand.

**Forward pass:** Data flows from the input to the output (left → right).

**Backward pass (backpropagation)**: Gradients are computed from the loss back to the input (right → left) by repeatedly applying the chain rule.

---

### Key Takeaways

- A neural network is a composition of many functions.
- The chain rule computes derivatives through these composed functions.
- Backpropagation repeatedly applies the chain rule.
- TensorFlow automatically performs these computations.
