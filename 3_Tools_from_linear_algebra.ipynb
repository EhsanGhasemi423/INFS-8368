For current purposes, you can think of a **vector** as a fixed-length array of scalars.

As with their code counterparts, we call these scalars the elements of the vector (synonyms include entries and components).

When vectors represent examples from real-world datasets, their values hold some real-world significance. For example, if we were training a model to predict the risk of a loan defaulting, we might associate each applicant with a vector whose components correspond to quantities like their income, length of employment, or number of previous defaults. Just as scalars are 0th-order tensors and vectors are 1st-order tensors, matrices are 2nd-order tensors. We denote matrices by bold capital letters (e.g., **X**, **Y**, and **Z**), and represent them in code by tensors with two axes.

A matrix can have any number of rows and columns. If the number of rows is the same as the number of columns, it is called a square matrix. To identify a specific value inside a matrix, we specify both its row and its column. For example, we might refer to the value in the second row and third column of the matrix.

We can convert any appropriately sized m * n tensor into an m*n matrix by passing the desired shape to **reshape**:

```python
A = tf.reshape(tf.range(6), (3, 2))
A
```

**Tensors**

A **tensor** is a generalization of scalars, vectors, and matrices to higher dimensions.

Tensors can have any number of axes (dimensions).

They are the primary data structure used in machine learning frameworks such as TensorFlow and PyTorch.

Scalars, vectors, and matrices are special cases of tensors:
  0th-order tensor: Scalar (single value)
  1st-order tensor: Vector (1D array)
  2nd-order tensor: Matrix (2D table)
  3rd-order tensor: Image (height × width × color channels)
  4th-order tensor: Collection (batch) of images

Higher-order tensors are created by adding more dimensions, allowing complex data (e.g., images, videos, and batches of data) to be represented efficiently.




```python
tf.reshape(tf.range(24), (2, 3, 4))
```

```python
A = tf.reshape(tf.range(6, dtype=tf.float32), (2, 3))
B = A  # No cloning of A to B by allocating new memory
A, A + B
```

The name Hadamard comes from the French mathematician Jacques Hadamard, who studied this type of operation.

```python
A * B
```

There is a simple function to calculate the **sum** of a **tensor’s elements**.

A tensor can have many dimensions (axes). Summing over all axes means adding every number in the tensor into one total.

```python
x = tf.range(3, dtype=tf.float32)
x, tf.reduce_sum(x)
```

```python
A.shape, tf.reduce_sum(A)
```

By default, tf.reduce_sum() adds all the elements in a tensor and returns a single value (a scalar). You can also specify an axis to sum along. When you sum along **axis=0**, TensorFlow **adds the values in each column**. Since the rows are combined into one value per column, the row dimension (axis 0) is removed from the output shape.

```python
A.shape, tf.reduce_sum(A, axis=0).shape
```

axis=0 is being removed after the summation.

```python
G = tf.constant([
    [1, 2, 3],
    [4, 5, 6]
])

print(G.shape)

print(tf.reduce_sum(G, axis=0))

print(tf.reduce_sum(G, axis=0).shape)
```

axis=1 sums the values in each row, producing one value per row.

```python
import tensorflow as tf

H = tf.constant([
    [2, 4, 6],
    [1, 3, 5]
])

print(H.shape)

print(tf.reduce_sum(H, axis=1))

print(tf.reduce_sum(H, axis=1).shape)
```

Reducing a matrix along both rows and columns via summation is equivalent to summing up all the elements of the matrix.

```python
tf.reduce_sum(A, axis=[0, 1]), tf.reduce_sum(A)  # Same as tf.reduce_sum(A)
```

A related quantity is the **mean**, also called the average. We calculate the mean by dividing the sum by the total number of elements. Because computing the mean is so common, it gets a dedicated **library function** that works analogously to sum.

```python
tf.reduce_mean(A), tf.reduce_sum(A) / tf.size(A).numpy()
```

Normally, reduce_sum() removes a dimension. If you set **keepdims=True**, TensorFlow keeps that dimension so the result is easier to use in later calculations, such as dividing the original matrix by its row sums.

keepdims=True **keeps the reduced dimension** instead of removing it.

```python
A.shape
```

```python
sum_A = tf.reduce_sum(A, axis=1, keepdims=True)
sum_A, sum_A.shape
```

A cumulative sum adds values one at a time, keeping a running total. Unlike reduce_sum(), it does not reduce the size of the tensor.

```python
I = tf.constant([
    [1, 2, 3],
    [4, 5, 6]
])

print(tf.cumsum(I, axis=0))
```

**Norms**

A norm measures the length (or magnitude) of a vector.

The **most common norm is the L2** norm (Euclidean norm), which is the straight-line distance from the origin to the vector.

It is the distance from the origin (0,0) to the tip of the vector.

Just like finding the hypotenuse of a right triangle, the L2 norm uses the Pythagorean theorem:

$$
\|\mathbf{x}\|_2 =
\sqrt{x_1^2 + x_2^2 + \cdots + x_n^2}
$$

```python
u = tf.constant([3.0, -4.0])
tf.norm(u)
```

**L1 Norm (Manhattan Distance)**

Definition: The L1 norm is the sum of the absolute values of a vector's elements.
$$
\|\mathbf{x}\|_1=\sum_{i=1}^{n}|x_i|
$$

The L1 norm is computed by taking the absolute value of each element and then summing them.

```python
tf.reduce_sum(tf.abs(u))
```

**Matrix Norms**


Just as vectors have norms, matrices also have norms.

The most common matrix norm is the Frobenius norm, which measures the size of a matrix.

$$
\|\mathbf{X}\|_F =
\sqrt{\sum_{i=1}^{m}\sum_{j=1}^{n} x_{ij}^{2}}
$$

How to compute it:
1. Square every element.
2. Add all the squared values.
3. Take the square root.

$$
X=
\begin{bmatrix}
1 & 2\\
3 & 4
\end{bmatrix}
$$

$$
\|\mathbf{X}\|_F
=
\sqrt{1^2+2^2+3^2+4^2}
=
\sqrt{30}
\approx5.48
$$

The **Frobenius** norm is the **L2 norm applied to all the elements of a matrix**.





```python
tf.norm(tf.ones((4, 9)))
```

## **Why Are Norms Important?**

Norms measure the size of a vector or matrix and the distance between predictions and actual values.

In machine learning and deep learning, norms are widely used to:

1. Measure prediction error (loss functions)
2. Measure similarity or distance between data points
3. Regularize models to reduce overfitting (L1 and L2 regularization)
4. Define optimization objectives that algorithms minimize during training

During training, machine learning models often minimize a norm (or a function based on a norm) to improve performance.

In deep learning, we are often trying to solve optimization problems: **maximize** the probability assigned to observed data; **maximize** the revenue associated with a recommender model; **minimize** the distance between predictions and the ground truth observations; **minimize** the distance between representations of photos of the same person while **maximizing** the distance between representations of photos of different people.

These **distances**, which constitute the objectives of deep learning algorithms, are often **expressed as norms**.
